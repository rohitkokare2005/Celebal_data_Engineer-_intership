Q4: Given a DataFrame df_sales, write a query to filter for rows where the region is 'West' and then group by product_category to find the average sale_amount. 

```python
df_sales.filter(col("_c2")=="West") \
.groupBy("_c3") \
.agg(avg(expr("try_cast(_c4 as double)")).alias("Average_Sale")) \
.show()
```

Q13: How do you use the .agg() function to calculate multiple statistics at once, such as the min, max, and mean of the price column? 

```python
df_sales.filter(col("_c0") != "user_id").agg(
min(expr("try_cast(_c4 as double)")).alias("Minimum"),
max(expr("try_cast(_c4 as double)")).alias("Maximum"),
avg(expr("try_cast(_c4 as double)")).alias("Average")
).show()
```

Q14: In the context of cleaning a dataset, what is the risk of using inferSchema=true when your source data contains messy or inconsistent date formats? 
ANS:
When source data contains inconsistent date formats, inferSchema=True may:
- 	Infer the wrong data type. 
- 	Treat dates as strings. 
- 	Produce null values during parsing. 
- 	Cause incorrect query results. 
Using an explicit schema is safer for production datasets.


Q15: Write a final processing pipeline that: 

Filters out duplicates. 

Fills null prices with 0. 

Groups by store_id to calculate total revenue. 

```python
result = df_sales \
.filter(col("_c0") != "user_id") \
.dropDuplicates(["_c0","_c1"]) \
.na.fill({"_c4":"0"}) \
.groupBy("_c12") \
.agg(sum(expr("try_cast(_c4 as double)")).alias("Total_Revenue"))

result.show()
```
